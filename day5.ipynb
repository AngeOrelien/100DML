```python
"""
    ------------------------------------------------------------------------------------------------------------
    100DML - Day 5: Manipulation de la Bibliotheque MATPLOTLIB
    ------------------------------------------------------------------------------------------------------------
    
    Matplotlib est une bibliothèque Python très populaire pour la création de graphiques et de visualisations de données. Elle offre 
    des outils puissants pour créer des graphiques 2D et 3D, des histogrammes, des diagrammes en barres, etc.
"""
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
x = np.linspace(0, 10, 1000) # Génère 1000 points entre 0 et 10
y = np.exp(x) # Calcule l'exponentielle de chaque point de x
```

```python
plt.plot(x, y) # Trace la courbe de y en fonction de x
plt.show()
```

```python
# changer la taille du graphique
plt.figure(figsize=(10, 5))
plt.plot(x, y)
```

```python
# changer le titre du graphique
plt.title("Fonction Exponentielle")
plt.xlabel('x')
plt.ylabel('Exp(x)')
plt.plot(x, y, c="red")
plt.show()
```

```python
# changer le type de ligne
plt.plot(x, y, c='green', ls='--')
```

```python
# ajouter la legende (label et legend)
plt.plot(x, y, c='green', ls='--', label='exp')
plt.legend()

# ajoute la grille
plt.grid()
```

```python
# un 2e graphique
z = 2 * y
plt.plot(x, y, c='green', ls='--', label='exp')
plt.plot(x, z, c='red', ls='-', label='2exp')

plt.legend()

# ajoute la grille
plt.grid()
```

```python
# zoomer sur une partie precise du graphique
plt.xlim(4, 8)
plt.ylim(0, 1000)
plt.plot(x, y, c='green', ls='--', label='exp')
plt.plot(x, z, c='red', ls='-', label='2exp')
```

```python
# ajouter les sous graphes dans les graphes
plt.subplot(2, 2, 1) # matrice nombre de graphes et ensuite index (ici 1)
plt.plot(x, y, c='red', label='exp')
plt.subplot(2, 2, 2) # matrice nombre de graphes et ensuite index (ici 2)
plt.plot(x, y, c='green', label='exp')
plt.subplot(2, 2, 3) # matrice nombre de graphes et ensuite index (ici 3)
plt.plot(x, y, c='blue', label='exp')
plt.subplot(2, 2, 4) # matrice nombre de graphes et ensuite index (ici 4)
plt.plot(x, y, c='orange', label='exp')

# ajouter l'espacement
plt.subplots_adjust(wspace=0.5)

# sauvegarder les graphiques en image

os.makedir('out')
plt.savefig('figure.png')
```

```python
# avoir le nuage de points
# 1. Identifier des tendances ou des relations entre deux variables
# 2. Détecter des corrélations (positives, négatives ou absence de corrélation)
# 3. Repérer des valeurs aberrantes (outliers)
# 4. Observer la distribution des données
# 5. Identifier des regroupements (clusters) potentiels
plt.scatter(y, y, c="red")
```

```python
# afficher l'histoigrame de distribution des valeurs
# La distribution sur un graphique (comme un histogramme) sert à:

# 1. **Visualiser la répartition des données** - Elle montre comment les valeurs sont distribuées sur l'ensemble de données.

# 2. **Identifier la forme des données** - Elle révèle si les données suivent une distribution normale (en cloche), sont asymétriques, bimodales, ou ont une autre forme particulière.

# 3. **Repérer les tendances centrales** - Elle permet de visualiser où se concentrent la majorité des valeurs.

# 4. **Détecter les valeurs aberrantes** - Elle aide à identifier les valeurs qui s'écartent significativement du reste des données.

# 5. **Comprendre la variabilité** - Elle montre l'étendue et la dispersion des données.

plt.hist(y)
```

```python
# afficher le boxplot pour voir les valeurs abherentes Le boxplot (ou boîte à moustaches) 
# est un outil statistique graphique qui sert à visualiser la distribution d'un ensemble de données.
# 1. **Identifier les valeurs aberrantes (outliers)** : Les points situés au-delà des "moustaches" sont considérés comme des valeurs aberrantes.
# 2. **Visualiser la dispersion des données** : La taille de la boîte et la longueur des moustaches indiquent l'étendue de la distribution.


# Une valeur aberrante (ou outlier en anglais) est une observation qui s'écarte significativement des autres observations d'un ensemble de données

plt.boxplot(y)
```
